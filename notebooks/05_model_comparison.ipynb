```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
results = {
    'Model': [
        'Linear Regression',
        'Random Forest',
        'XGBoost',
        'LSTM'
    ],

    'MAE': [
        5.7608,
        6.7108,
        7.0787,
        10.1034
    ],

    'RMSE': [
        8.0601,
        9.2664,
        9.5836,
        13.7530
    ],

    'R2 Score': [
        0.9691,
        0.9591,
        0.9563,
        0.9132
    ]
}

results_df = pd.DataFrame(results)

results_df
```

```python
plt.figure(figsize=(10,5))

sb.barplot(
    x='Model',
    y='R2 Score',
    data=results_df
)

plt.title('Model Comparison Based on R2 Score')

plt.ylabel('R2 Score')

plt.xticks(rotation=10)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sb.barplot(
    x='Model',
    y='MAE',
    data=results_df
)

plt.title('Model Comparison Based on MAE')

plt.ylabel('MAE')

plt.xticks(rotation=10)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sb.barplot(
    x='Model',
    y='RMSE',
    data=results_df
)

plt.title('Model Comparison Based on RMSE')

plt.ylabel('RMSE')

plt.xticks(rotation=10)

plt.show()
```

```python
overfitting_results = {
    'Model': [
        'Linear Regression',
        'Random Forest',
        'XGBoost',
        'LSTM'
    ],

    'Train R2': [
        0.9974,
        0.9996,
        0.9996,
        0.9930
    ],

    'Test R2': [
        0.9691,
        0.9591,
        0.9563,
        0.9132
    ]
}

overfit_df = pd.DataFrame(overfitting_results)

overfit_df
```

```python
overfit_df.set_index('Model').plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Train vs Test R2 Scores')

plt.ylabel('R2 Score')

plt.xticks(rotation=10)

plt.show()
```

Final Conclusions
Linear Regression achieved the best overall performance on the Tesla stock dataset.
Random Forest and XGBoost also performed strongly but showed slight overfitting.
LSTM successfully captured sequential trends but struggled with sudden price spikes.
Traditional machine learning models outperformed the deep learning model for this dataset.
Feature engineering significantly improved prediction performance.
